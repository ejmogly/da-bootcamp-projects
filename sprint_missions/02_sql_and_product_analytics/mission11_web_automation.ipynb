# HTML에서 중요한 데이터 추출하기

### HTML 구조 이해하기

1. HTML의 각 요소들은 모두 `태그`로 지정이 된다.

2. HTML 태그들은 `트리 구조`를 갖고 있다.
    - 부모 태그
    - 자식 태그

    ex.
    ```html
        <html>
            <head>
                <title> </title>
            </head>
        </html>
    ```

    ex.
    - `<p></p>` : paragraph 준말, 하나의 문단을 나타내는 태그
    - `<b></b>`
    - `<hr>` : 수평선 태그, 끝 태그X, 내용물이 없기 때문에.
    - `<ul><\ul>` : unordered list의 준말, 순서가 없는 목록을 나타내는 태그
    - `<ol><\ol>` : ordered list의 준말, 순서가 있는 목록을 나타내는 태그
    - `<li><\li>` : list item의 준말, 목록의 각 항목들을 나타내는 태그

3. HTML 태그는 `속성`을 가질 수 있다.
    - 속성을 통해 태그의 정보를 불러온다.
    ex. `<img src="이미지의 파일 경로">`
        - `<img><\img>` 태그의 src(source) 속성은 불러올 이미지의 경로를 입력 받는다.
    ex2. `<a href="웹 링크"><\a>`
        - `<a><\a>` 태그의 속성 href(hyperlink reference)는 이동할 웹 주소를 입력 받는다.

### 1. **HTML 기본 구조**
- **`<!DOCTYPE html>`**: HTML5 문서를 선언하는 태그. (문서 가장 먼저 선언)
- **`<html>`**: HTML 문서의 시작과 끝을 나타냄.
- **`<head>`**: 메타데이터(페이지 정보, 스타일, 스크립트 등)를 포함.
- **`<title>`**: 브라우저 탭, 방문기록에 표시되는 페이지 제목.

```html
<!DOCTYPE html>
<html>
  <head>
    <title>Sample Website</title>
  </head>
  <body>
    <!-- Body content here -->
  </body>
</html>
```

## HTML `style` 속성에서 사용된 CSS 속성(프로퍼티) 설명

- HTML의 `style` 속성 안에는 다양한 `CSS 속성(프로퍼티)`들이 들어가 있으며, 이 속성들은 `웹 요소의 스타일을 지정`하는 데 사용됩니다. 아래는 위 코드에서 사용된 각 CSS 속성에 대한 설명

1. `color`
- **설명**: 텍스트의 색상을 지정
- **예시**: `color: royalblue;` → 텍스트 색상을 **royalblue**로 설정

2. `text-align`
- **설명**: 텍스트의 정렬 방식을 지정
- **값**: `left`, `right`, `center` 등
- **예시**: `text-align: center;` → 텍스트를 **가운데 정렬**

3. `background-color`
- **설명**: 요소의 배경 색상을 지정
- **예시**: `background-color: lightblue;` → 요소의 배경 색상을 **연한 파랑**으로 설정

4. `font-style`
- **설명**: 텍스트의 기울임체 스타일을 지정
- **값**: `normal`, `italic`, `oblique` 등
- **예시**: `<i>` 태그로 텍스트를 감싸면 `font-style: italic;`이 적용

5. `font-weight`
- **설명**: 텍스트의 굵기를 설정
- **값**: `normal`, `bold`, `lighter`, 숫자 값(`100` ~ `900`)
- **예시**: `<b>` 태그를 사용하면 `font-weight: bold;`가 적용

6. `border`
- **설명**: 요소에 테두리를 설정할 수 있음
- **예시**: `<hr>` 태그는 기본적으로 수평선(border)의 역할

# 스타일 분리하기

HTML에 `style` 속성을 포함해 직접 작성하면 다음과 같은 `문제점들` 발생

1. `가독성 저하`: 스타일 코드가 길어지면, HTML 코드 자체가 복잡해지고 읽기 어려워짐
2. `관리의 어려움(유지 보수성)`: 여러 태그에 같은 스타일을 적용하려면 일일이 `style` 속성을 넣어줘야 하고, 변경할 때에도 동일하게 하나씩 수정해야 함

- 이 문제를 해결하기 위해 `HTML과 CSS를 분리해 스타일을 관리할 수 있음.` HTML은 웹페이지의 **내용물**을 정의하고, CSS는 스타일적 속성을 담당. 이렇게 분리하여 HTML과 CSS를 연결해주는 방식을 사용

      

-	여기서 `href는 CSS 파일의 경로`. HTML 파일과 CSS 파일은 `같은 폴더 안`에 있어야 원활히 연결

### 1. CSS 파일 생성 및 연결

1. **CSS 파일 생성**:
   - 별도의 CSS 파일(예: `style.css`)을 생성
   
2. **HTML과 CSS 연결**:
   - HTML 파일의 `<head>` 태그 안에 다음과 같은 코드를 추가하여 CSS 파일을 HTML과 연결:
      ```html
      <head>
         <link rel="stylesheet" href="style.css">
      </head>

      ```

### 2. CSS 선택자와 스타일 코드

- CSS에서 `선택자`는 `어떤 HTML 요소`에 스타일을 적용할지를 결정. 그리고 스타일 코드는 해당 선택자에 적용할 구체적인 스타일 속성들을 정의함

- 기본 문법:

  ```css
  /* 선택자 */ {
    /* 스타일 코드 */
  }
  ```

- h2 태그에 대한 스타일을 적용하려면 다음과 같이 작성

  ```css
  h2 {
    color: royalblue;
    text-align: center;
  }

## 다양한 선택자 종류
##### 1. 태그 이름 선택자

- HTML `태그 이름`을 사용해 특정 태그 전체에 스타일을 적용


	-	예: h2, p, div

    - 문법
        ```css
        tagname {
            /* 스타일 코드 */
        }
        ```

    - 예시
        ```css
        h2 {
            color: royalblue;
        }
        ```


##### 2.	ID 선택자 (오직 1개)

- ID는 문서에서 `유일한 요소`에 스타일을 적용할 때 사용
- 하나의 문서에서 ID 값은 `하나만 사용`

	- 문법:
		- html
			```html
			<h2 id="id-name">메인 제목</h2>
			```
		- css
			```css
			#id-name {
			/* 스타일 코드 */
			}
			```

	- 예시:
		```css
		#id-name {
			color: blue;
		}
		```
		- id가 id-name인 요소에 스타일이 적용
  

##### 3.	클래스 선택자 (여러 다른 태그들)

- Class는 `여러 HTML 태그에 공통`으로 적용
- `띄어쓰기로 여러 클래스 이름`을 사용

  -	예: .highlight, h2.small.subtitle, .subtitle.small, h2.subtitle (다양한 조합, 순서 관계없음)

  - 문법:
    - html
      ```html
      <h2 class="class-name">첫 번째 제목</h2>
      <h2 class="class-name class-name2">두 번째 제목</h2>
      ```
    - css
      ```css
      .class-name {
          /* 스타일 코드 */
      }
      ```
      ```css
      .class-name.class-name2 {
          /* 스타일 코드 */
      }
      ```
  - 예시:
    ```css
    .codeit {
    font-weight: bold;
    }
    ```
    - class가 codeit인 태그에 스타일이 적용

##### 4. 속성 선택자

- id와 class 외에도 `HTML 속성`을 선택자로 사용. 특정 속성 값을 가진 요소만 선택
    - 예: `[attr="value"]`

    - 문법
        - html
            ```html
            <a attr="value">코드잇 홈페이지</a>
            ```

        - css
            ```css
            [attr="value"] {
                /* 스타일 코드 */
            }
            ```
            ```css
            a[attr="value"] {
                /* 스타일 코드 */
            }
            ```

    - 예시:
        ```css
        [href="/"] {
            text-decoration: none;
        }
        ```
        - href 속성 값이 /인 요소에 스타일이 적용


#### 5. 직속 자식 선택자 (`>`)

- 선택한 요소의 `직속 자식만` 선택할 때 사용

    - 문법:
        ```css
        selector1 > selector2 {
            /* 스타일 코드 */
        }
        ```

    - 예시:
        ```css
        li > i {
            color: green;
        }
        ```

        - li 태그의 직속 자식인 i 태그에만 스타일이 적용


#### 6. 전체 선택자 (*)

별표(*)는 `모든 태그를 선택`할 때 사용

- 예시
    ```css
    li * {
        color: green;
    }
    ```
        - li 태그 안에 있는 모든 태그에 스타일이 적용



### 7. 요약

-	`태그 이름 선택자`: 특정 태그를 선택.
-	`ID 선택자:` 유일한 요소를 선택.
-	`Class 선택자:` 여러 요소에 동일한 스타일을 적용.
-	`속성 선택자:` 특정 속성 값을 가진 요소를 선택.
-	`OR 선택자 (,):` 여러 선택자를 조합해 하나라도 맞으면 적용.
-	`AND 선택자 (.):` 특정 클래스에 속하는 요소를 선택.
-	`일반 자식 선택자` (띄어쓰기): 모든 자식 요소를 선택.
-	`직속 자식 선택자` (>): 직속 자식 요소만 선택.

## selenium wait
Selenium은 웹과 직접 상호작용하기 때문에, 웹 요소가 모두 로드될 때까지 기다리는 작업이 필요

---

1. `implicitly_wait()`
- 한 번 설정하면 Selenium 작업 전체에 적용.
- 특정 요소가 나타날 때까지 최대 N초 기다리는 방식
    - 단순히 N초를 기다리는 것이 아니라, 해당 요소가 나타날 때까지 계속 시도

    ```python
        driver.implicitly_wait(3)
    ```

2. `time.sleep()`

-	정확히 N초를 기다림
-	새로운 창이 로드될 때 안전하게 사용
    -	예: 로그인 창 로드, 페이지 로딩 등

    ```python
    from time import sleep
    sleep(1)
    ```


`````python
# 라이브러리 임포트
from time import sleep

from selenium import webdriver
from selenium.webdriver.common.by import By
### explicit wait ###
from selenium.webdriver.support import expected_conditions as EC
from selenium.webdriver.support.ui import WebDriverWait
###

# 크롬 드라이버 생성
driver = webdriver.Chrome()
wait = WebDriverWait(driver, 3)

# 사이트 접속하기
driver.get('https://workey.codeit.kr/costagram')

# 로그인 버튼 클릭하기
# 최대 3초, 조건에 대한 인자는 튜플, 길어보여도 우리는 조건과 선택자만 바꿀 것!(암기X)
login_link = wait.until(EC.element_to_be_clickable( (By.CSS_SELECTOR, 'a.top-nav__login-link') ))
login_link.click()

# ID와 패스워드 입력하기
login_input = wait.until(EC.visibility_of_element_located( (By.CSS_SELECTOR, '.login-container__login-input') ))
login_input.send_keys('codeit')

password_input = wait.until(EC.visibility_of_element_located( (By.CSS_SELECTOR, '.login-container__password-input') ))
password_input.send_keys('datascience')

# 로그인 버튼 클릭하기
login_btn_link = wait.until(EC.element_to_be_clickable( (By.CSS_SELECTOR, '.login-container__login-button') ))
login_btn_link.click()

# 5초 기다리기 - 너무 빨리 닫혀 확인이 어렵기 떄문.
sleep(5)

# 크롬 드라이버 종료 - 항상 종료를 해주는 것이 좋다. (#메모리 누수 방지 #브라우저 실행 충돌 방지)
driver.quit()
`````

### Selenium Action Chain

- Selenium에서 Action Chain은 사용자의 마우스, 키보드 동작을 `사슬처럼 이어서 실행`하는 것을 의미
- 이를 통해 `더블 클릭, 드래그, 우클릭` 등 복잡한 동작을 자동화
- 복잡한 시나리오가 있을 때 유용하지만, `대부분의 작업은 클릭과 입력만으로도 충분`

---

1. Action Chain 사용 방법

- 연결하는 방식: 한 줄로 이어서 실행
- 나열하는 방식: 액션을 각각 나열하여 실행

    <br>

    1-1. `연결하는 방식`

    ```python
    from selenium.webdriver.common.action_chains import ActionChains

    ActionChains(driver)
        .send_keys_to_element(id_box, 'codeit')
        .send_keys_to_element(pw_box, 'datascience')
        .click(login_button)
        .perform()
    ```

    1-2. `나열하는 방식`

    ```python
    from selenium.webdriver.common.action_chains import ActionChains

    actions = ActionChains(driver) # 인스턴스 선언
    actions.send_keys_to_element(id_box, 'codeit') # 메서드 나열
    actions.send_keys_to_element(pw_box, 'datascience')
    actions.click(login_button)
    actions.perform()
    ```

    - 두 방식 중 편한 방법을 선택하여 사용하면 됩니다.

2. Action Chain 기능 여러가지
    - parameter가 None인 건, 선택적 사용 가능하다는 뜻.

	1.	`클릭`: 특정 요소를 클릭하거나, 요소가 없을 경우 현재 마우스 위치를 클릭

        ```python
        .click(element=None)
        ```
    
    2. `오른쪽 클릭`: 특정 요소를 오른쪽 클릭하거나, 요소가 없을 경우 현재 마우스 위치에서 우클릭

        ```python
        .context_click(element=None)
        ```
    
    3. `더블 클릭`: 특정 요소를 더블 클릭하거나, 요소가 없을 경우 현재 마우스 위치에서 더블 클릭

        ```python
        .double_click(element=None)
        ```
    
    4. `드래그 앤 드롭`: source 요소를 클릭해서 target 요소로 드래그 후 드롭 (source -> target)

        ```python
        .drag_and_drop(source, target)
        ```
    5. `마우스 이동`: 특정 요소까지 마우스를 이동

        ```python
        .move_to_element(element)
        ```
    
    6. `키보드 입력`: 특정 요소에 키보드 입력

        ```python
        .send_keys(keys)
        .send_keys_to_element(element, keys) # 입력 대상 요소를 지정
        ```

    7. `동작 중지`: 일정 시간 동안 동작을 멈춤 (sleep 대신)

        ```python
        .pause(seconds)
        ```


## selenium에서 자바스크립트 사용
- execute_script() 메소드로 직접 자바스크립트 명령어 입력

`````python
from time import sleep
from selenium import webdriver

driver = webdriver.Chrome()
driver.implicitly_wait(3)

driver.get('https://workey.codeit.kr/music')
sleep(1)

# 첫 페이지 로딩시 웹페이지 높이 가져오기
### document.body.scrollHeight
### 반환시 return 붙여주기

last_height = driver.execute_script("return document.body.scrollHeight")

# 끝까지 스크롤링 반복
while True:
    # 첫 스크롤링
    driver.execute_script("window.scrollTo(0, document.body.scrollHeight)")

    # 새로운 페이지가 로딩될 때까지 대기
    sleep(0.5)

    # 새로운 내용이 로딩되었는지 확인
    new_height = driver.execute_script("return document.body.scrollHeight")

    if new_height == last_height: # 끝까지 다 내려왔다는 의미!
        break
    last_height = new_height

sleep(5)
driver.quit()
`````

# selenium과 beautifulsoup 같이 사용


- Selenium :
    - `사용자 동작(클릭, 스크롤, 로그인 등)`을 흉내내고 자동화할 수 있음.
    - 웹 페이지에서 `동적 컨텐츠`를 제어하는 데 유리 (스크롤, Ajax 로딩 등)
    - 하지만, 많은 양의 데이터를 처리하기엔 속도가 느릴 수 있음.

- BeautifulSoup :
    - `정적인 HTML 코드`를 빠르게 분석하고 필요한 데이터를 추출할 수 있음.
    - 복잡한 HTML 구조에서도 `효율적으로 데이터 추출`이 가능
    - 하지만, 사용자 동작을 수행하는 기능은 없습니다.

#### 왜 두 가지를 함께 사용?

- Selenium을 사용하면 동적 컨텐츠를 제어할 수 있음.
    - 예를 들어, 스크롤을 통해 새로운 데이터 로딩이나 로그인 같은 동작이 가능
- BeautifulSoup을 사용하면 정적 HTML 코드에서 데이터를 빠르게 추출할 수
- `둘의 조합`을 통해 `동적 웹 페이지에서도 효율적으로 데이터를 수집`할 수 있음.

---

#### HTML 코드 추출: `driver.page_source`

- Selenium을 통해 필요한 동작(예: 스크롤)을 완료한 후, `HTML 코드를 추출`하는 방법은 간단

    ```python
    # 스크롤 완료 후 HTML 코드 추출
    music_page = driver.page_source
    ```

    -	driver.page_source 로 현재 웹 페이지의 전체 HTML 코드를 가져올 수 있음.
    -	그 후, BeautifulSoup을 사용하여 HTML 데이터를 파싱하고 필요한 정보를 추출


## [참고] 명령어 비교
- 웹 스크래핑을 할 때, Beautiful Soup과 Selenium은 모두 사용 가능. 하지만, 두 도구의 역할이 다르며, `때로는 이 둘을 함께 사용하는 것이 더 효과적`

| Beautiful Soup   | Selenium              |
|----------------------|---------------------------|
| `soup.select_one()`   | `driver.find_element()`    |
| `soup.select()`       | `driver.find_elements()`   |
| Tag               | WebElement            |
| `tag.select()`        | `web_element.find_elements()` |
| `tag.get_text()`      | `web_element.text`        |
| `tag.strings()`       | 지원 안 됨             |
| `tag.stripped_strings()` | 지원 안 됨        |
| `tag['attr']`         | `web_element.get_attribute('attr')` |

- 웹 드라이버 생성 후 코스타그램 접속
- 코스타그램 로그인(ID : codeit, PW : datascience)
- 웹 페이지 끝까지 스크롤
- 각 썸네일(포스팅)을 클릭하고, 창이 뜨면 닫기 버튼 클릭
- 웹 드라이버 종료

```python
import time
from selenium import webdriver

# 웹 드라이버 설정 / 스크롤 참고

# 현재 scrollHeight 가져오기
last_height = driver.execute_script("return document.body.scrollHeight")

while True:
    # scrollHeight까지 스크롤
    driver.execute_script("window.scrollTo(0, document.body.scrollHeight);")

    # 새로운 내용 로딩될때까지 기다림
    time.sleep(0.5)

    # 새로운 내용 로딩됐는지 확인
    new_height = driver.execute_script("return document.body.scrollHeight")
    if new_height == last_height:
        break
    last_height = new_height
```
