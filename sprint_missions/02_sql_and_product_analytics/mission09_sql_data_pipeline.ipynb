```python
import glob
import pandas as pd
import koreanize_matplotlib
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
lst = glob.glob('*.csv')
lst
```

```python
data_description = pd.read_csv('data_description.csv') # 위 컬럼명세 데이터 정리한 것
# 오타수정
data_description['col']= data_description['col'].str.replace('product_name_length','product_name_lenght')
data_description['col']= data_description['col'].str.replace('product_description_length','product_description_lenght')
data_description.head()
```

```python
files = {}
for file_path in lst:
    if 'data_description' not in file_path:
        file_name =file_path.split('.')[0]
        globals()[file_name] = pd.read_csv(file_path) # 파일명을 변수명으로
        files[file_name] = {'raw':eval(file_name)}
```

```python
for file_name in lst:

    name = file_name.split('.')[0]
    if name != 'data_description':
        df   = files[name]['raw']
        map_col = {k:v for k,v in data_description[data_description.file.str.contains(name)][['col','desc']].values}
        if name =='product_category_name_translation':
            map_col['product_category_name'] = '카테고리이름'
            map_col['product_category_name_english'] = '영어이름'

        display_df = df.copy()
        columns = pd.MultiIndex.from_arrays([list(display_df.columns),[map_col[x] for x in list(display_df.columns)]])
        display_df.columns = columns
        files[name]['display'] = display_df
        print(name)
        display(display_df.head())
        print(display_df.shape)
        print('########################################################################\n\n')
```

```python
def get_display_columns(name):
    display(files[name]['display'].head(1))
```

### 데이터 전처리
1. customer_id / customer_unique_id 합치기 -> customer_unique_id 값으로만 데이터 분석 진행
2. order_status -> delivered 상태인 경우만
3. 시간은 order_purchase_timestamp로만 판단
4. 총 주문금액 컬럼과 주문 상품 리스트 추가 / 결제 방식의 금액과 횟수를 dict으로 추가

```python
customer_dict = {k:v for k,v in olist_customers_dataset[['customer_id','customer_unique_id']].values}
olist_orders_dataset['customer_unique_id'] = olist_orders_dataset['customer_id'].map(customer_dict)
order_data = olist_orders_dataset[olist_orders_dataset.order_status =='delivered'][['order_id','customer_unique_id','order_purchase_timestamp']].reset_index(drop=True)

order_payment = olist_order_payments_dataset.groupby(['order_id'])['payment_value'].sum().reset_index()
order_data['total_payments_values'] = order_data['order_id'].map({k:v for k,v in order_payment.values})
order_data = order_data.dropna().reset_index(drop=True) # 1개 행에서 결측치 발생
payments_data =olist_order_payments_dataset.groupby(['order_id','payment_type'])['payment_value'].agg(['sum','size']).reset_index()
```

```python
# 시간좀 오래걸림
dicts = {}

for order_id in payments_data.order_id.unique():
    target_df = payments_data[payments_data.order_id ==order_id]

    payments_values = {k:v for k,v in target_df[['payment_type','sum']].values}
    payments_method_size = {k:v for k,v in target_df[['payment_type','size']].values}
    payments_method_unique = list(target_df['payment_type'].values)


    dicts[order_id] = {
        'payments_values' :payments_values,
        'payments_method_size' :payments_method_size,
        'payments_method_unique':payments_method_unique
    }
```

# Retention
- DAU / MAU
- 2회 이상 구매자들에 대해 구매 코호트 분석
- 구매 클래식 retention

```python
dau = order['order_purchase_timestamp'].dt.date.value_counts().sort_index()
mau = dau.rolling(30).sum().dropna()

(dau/mau).dropna().plot()
plt.title('DAU/MAU')
plt.tight_layout()
plt.show()
```

```python
## 클래식 리텐션 최소 구매일자를 0일차, n일차에 구매했을 경우 카운트
```

```python
first_order = order[['customer_unique_id','order_purchase_timestamp']].sort_values(['customer_unique_id','order_purchase_timestamp']).drop_duplicates('customer_unique_id',keep='first').reset_index(drop=True)
first_order = first_order.rename(columns ={'order_purchase_timestamp':"first_order_time"})
order_retention = pd.merge(order[['customer_unique_id','order_purchase_timestamp']],first_order)
```

```python
order_retention['order_purchase_timestamp_date'] = pd.to_datetime(order_retention['order_purchase_timestamp'].dt.date)
order_retention['first_order_time_date'] = pd.to_datetime(order_retention['first_order_time'].dt.date)
```

```python
order_retention['diff_time'] = (order_retention['order_purchase_timestamp_date'] - order_retention['first_order_time_date']).dt.days
```

```python
(order_retention['diff_time'].value_counts().sort_index() / order_retention['diff_time'].value_counts()[0]) # 1일 구매리텐션이 0.05%
```

```python
(order_retention['diff_time'].value_counts().sort_index() / order_retention['diff_time'].value_counts()[0]).plot()
plt.title('order retention')
plt.xlabel('diff_days')
plt.show()
```

재구매 안한 비율이 높기 때문에 거의 0%에 근접 -> 2회 이상 구매한 유저들에 대한 구매리텐션 확인

```python
multiple_buy = order_retention.customer_unique_id.value_counts()[order_retention.customer_unique_id.value_counts() >=2].index
filter_multiple_buy = order_retention[order_retention.customer_unique_id.isin(multiple_buy)]


(filter_multiple_buy['diff_time'].value_counts().sort_index() / filter_multiple_buy['diff_time'].value_counts()[0]).plot()
plt.title('order retention')
plt.xlabel('diff_days')
plt.show()
```

```python
(filter_multiple_buy['diff_time'].value_counts().sort_index() / filter_multiple_buy['diff_time'].value_counts()[0]) # 1일 구매리텐션이 1%
```

30일 구매 리텐션

```python
filter_multiple_buy = filter_multiple_buy.reset_index(drop=True)
filter_multiple_buy['diff_month'] = (filter_multiple_buy['diff_time'] + 29 ) // 30
```

```python
(filter_multiple_buy['diff_month'].value_counts().sort_index() / filter_multiple_buy['diff_month'].value_counts()[0]) # 1달 구매 리텐션 17%
```

```python
(filter_multiple_buy['diff_month'].value_counts().sort_index() / filter_multiple_buy['diff_month'].value_counts()[0]).plot()
plt.title('order month retention')
plt.xlabel('diff_month')
plt.show()
```
