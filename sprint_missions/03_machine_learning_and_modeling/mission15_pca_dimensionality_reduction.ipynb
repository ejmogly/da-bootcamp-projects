# 추가 분석: 신용카드 고객 세분화를 위한 파생변수 생성 및 군집분석

아래 셀부터는 결측치 처리가 끝난 `df`를 기준으로 진행합니다.  
분석 목표는 단순 EDA가 아니라 **고객 행동 기반 파생변수 생성 → 군집분석 → 군집별 마케팅 전략 도출**입니다.

핵심 흐름은 다음과 같습니다.

1. 원본 데이터 보존 및 ID 제거
2. 고객 행동 기반 파생변수 생성
3. EDA 및 상관관계 확인
4. 이상치 완화, 로그변환, 표준화
5. Elbow / Silhouette / DBI / CH Score로 군집 수 검토
6. K-Means 군집분석
7. PCA 2차원 시각화
8. 군집별 프로파일링
9. 군집별 마케팅 전략 제안


```python
# ============================================================
# 1. 기본 라이브러리 및 분석 환경 설정
# ============================================================

import numpy as np
import pandas as pd

import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score, davies_bouldin_score, calinski_harabasz_score
from sklearn.decomposition import PCA

import warnings
warnings.filterwarnings('ignore')

plt.rcParams['figure.figsize'] = (10, 6)
plt.rcParams['axes.unicode_minus'] = False

RANDOM_STATE = 42

# 결측치 처리가 끝난 df를 기준으로 시작
df_raw = df.copy()

print('현재 데이터 크기:', df_raw.shape)
display(df_raw.head())
```

```python
# ============================================================
# 2. 분석용 데이터 준비: ID 컬럼 제거 및 숫자형 변환
# ============================================================

# df.reset_index()를 사용한 경우 생기는 index 컬럼은 분석에 불필요하므로 제거
drop_cols = []

if 'index' in df_raw.columns:
    drop_cols.append('index')

# 고객 ID는 모델링 변수로 사용하지 않음
id_col = None
if 'CUST_ID' in df_raw.columns:
    id_col = df_raw['CUST_ID'].copy()
    drop_cols.append('CUST_ID')

df_base = df_raw.drop(columns=drop_cols, errors='ignore').copy()

# 숫자형 변환
df_base = df_base.apply(pd.to_numeric, errors='coerce')
```

```python
# ============================================================
# 3. 안전한 나눗셈 함수 정의
# ============================================================

def safe_divide(numerator, denominator):
    """
    비율 파생변수 생성 시 0으로 나누는 문제를 방지하는 함수
    - denominator가 0이면 NaN 처리 후 0으로 대체
    - inf, -inf도 0으로 대체
    """
    numerator = pd.to_numeric(numerator, errors='coerce').fillna(0)
    denominator = pd.to_numeric(denominator, errors='coerce').replace(0, np.nan)

    result = numerator / denominator
    result = result.replace([np.inf, -np.inf], np.nan).fillna(0)

    return result
```

```python
# ============================================================
# 4. 파생변수 생성
# ============================================================
# 목적: 단순 금액 변수보다 고객의 행동 패턴을 더 잘 설명하는 비율/점수 변수 생성

df_feature = df_base.copy()

# 1) 한도 사용률: 신용한도 대비 미상환잔액
df_feature['LIMIT_USAGE_RATE'] = safe_divide(
    df_feature['BALANCE'],
    df_feature['CREDIT_LIMIT']
)

# 2) 구매 비율: 신용한도 대비 구매금액
df_feature['PURCHASE_TO_LIMIT'] = safe_divide(
    df_feature['PURCHASES'],
    df_feature['CREDIT_LIMIT']
)

# 3) 현금서비스 비율: 신용한도 대비 현금서비스 금액
df_feature['CASH_ADVANCE_TO_LIMIT'] = safe_divide(
    df_feature['CASH_ADVANCE'],
    df_feature['CREDIT_LIMIT']
)

# 4) 결제 비율: 미상환잔액 대비 결제금액
df_feature['PAYMENT_TO_BALANCE'] = safe_divide(
    df_feature['PAYMENTS'],
    df_feature['BALANCE'] + 1
)

# 5) 최소결제 부담률: 미상환잔액 대비 최소결제금액
df_feature['MIN_PAYMENT_PRESSURE'] = safe_divide(
    df_feature['MINIMUM_PAYMENTS'],
    df_feature['BALANCE'] + 1
)

# 6) 일시불 구매 비중
df_feature['ONEOFF_PURCHASE_RATIO'] = safe_divide(
    df_feature['ONEOFF_PURCHASES'],
    df_feature['PURCHASES'] + 1
)

# 7) 할부 구매 비중
df_feature['INSTALLMENT_PURCHASE_RATIO'] = safe_divide(
    df_feature['INSTALLMENTS_PURCHASES'],
    df_feature['PURCHASES'] + 1
)

# 8) 현금서비스 의존도: 구매 + 현금서비스 중 현금서비스 비중
df_feature['CASH_ADVANCE_DEPENDENCY'] = safe_divide(
    df_feature['CASH_ADVANCE'],
    df_feature['PURCHASES'] + df_feature['CASH_ADVANCE'] + 1
)

# 9) 구매 중심도: 구매 + 현금서비스 중 구매 비중
df_feature['PURCHASE_DEPENDENCY'] = safe_divide(
    df_feature['PURCHASES'],
    df_feature['PURCHASES'] + df_feature['CASH_ADVANCE'] + 1
)

# 10) 거래 1건당 평균 구매금액
df_feature['AVG_PURCHASE_AMOUNT_PER_TRX'] = safe_divide(
    df_feature['PURCHASES'],
    df_feature['PURCHASES_TRX']
)

# 11) 현금서비스 1건당 평균 금액
df_feature['AVG_CASH_ADVANCE_AMOUNT_PER_TRX'] = safe_divide(
    df_feature['CASH_ADVANCE'],
    df_feature['CASH_ADVANCE_TRX']
)

# 12) 구매 활동성 점수
df_feature['PURCHASE_ACTIVITY_SCORE'] = (
    df_feature['PURCHASES_FREQUENCY'] +
    df_feature['ONEOFF_PURCHASES_FREQUENCY'] +
    df_feature['PURCHASES_INSTALLMENTS_FREQUENCY']
) / 3

# 13) 잔액 부담/리볼빙 위험 점수
# 한도 사용률이 높고 전액결제 비율이 낮을수록 위험도가 높다고 해석
df_feature['REVOLVING_RISK_SCORE'] = (
    df_feature['LIMIT_USAGE_RATE'] * (1 - df_feature['PRC_FULL_PAYMENT'])
)

# 14) 결제 건전성 점수
# 전액결제 비율과 잔액 대비 결제 비율을 함께 반영
df_feature['PAYMENT_HEALTH_SCORE'] = (
    df_feature['PRC_FULL_PAYMENT'] + df_feature['PAYMENT_TO_BALANCE']
)

# inf, NaN 최종 정리
df_feature = df_feature.replace([np.inf, -np.inf], np.nan).fillna(0)

print('파생변수 생성 후 데이터 크기:', df_feature.shape)
display(df_feature.head())
```

```python
# ============================================================
# 5. 파생변수 요약 확인
# ============================================================

derived_cols = [
    'LIMIT_USAGE_RATE',
    'PURCHASE_TO_LIMIT',
    'CASH_ADVANCE_TO_LIMIT',
    'PAYMENT_TO_BALANCE',
    'MIN_PAYMENT_PRESSURE',
    'ONEOFF_PURCHASE_RATIO',
    'INSTALLMENT_PURCHASE_RATIO',
    'CASH_ADVANCE_DEPENDENCY',
    'PURCHASE_DEPENDENCY',
    'AVG_PURCHASE_AMOUNT_PER_TRX',
    'AVG_CASH_ADVANCE_AMOUNT_PER_TRX',
    'PURCHASE_ACTIVITY_SCORE',
    'REVOLVING_RISK_SCORE',
    'PAYMENT_HEALTH_SCORE'
]

print('파생변수 기초통계량')
display(df_feature[derived_cols].describe().T)
```

```python
# ============================================================
# 6. EDA: 핵심 파생변수 분포 확인
# ============================================================

eda_dist_cols = [
    'LIMIT_USAGE_RATE',
    'PURCHASE_TO_LIMIT',
    'CASH_ADVANCE_TO_LIMIT',
    'PAYMENT_TO_BALANCE',
    'CASH_ADVANCE_DEPENDENCY',
    'REVOLVING_RISK_SCORE'
]

for col in eda_dist_cols:
    plt.figure(figsize=(8, 4))
    sns.histplot(df_feature[col], bins=50, kde=True)
    plt.title(f'Distribution of {col}')
    plt.xlabel(col)
    plt.ylabel('Count')
    plt.show()
```

```python
# ============================================================
# 7. EDA: 주요 변수 간 관계 시각화
# ============================================================

# 구매금액 vs 현금서비스
plt.figure(figsize=(8, 6))
sns.scatterplot(data=df_feature, x='PURCHASES', y='CASH_ADVANCE', alpha=0.4)
plt.title('PURCHASES vs CASH_ADVANCE')
plt.show()

# 신용한도 vs 미상환잔액
plt.figure(figsize=(8, 6))
sns.scatterplot(data=df_feature, x='CREDIT_LIMIT', y='BALANCE', alpha=0.4)
plt.title('CREDIT_LIMIT vs BALANCE')
plt.show()

# 한도 사용률 vs 전액결제 비율
plt.figure(figsize=(8, 6))
sns.scatterplot(data=df_feature, x='LIMIT_USAGE_RATE', y='PRC_FULL_PAYMENT', alpha=0.4)
plt.title('LIMIT_USAGE_RATE vs PRC_FULL_PAYMENT')
plt.show()

# 현금서비스 의존도 vs 결제 건전성
plt.figure(figsize=(8, 6))
sns.scatterplot(data=df_feature, x='CASH_ADVANCE_DEPENDENCY', y='PAYMENT_HEALTH_SCORE', alpha=0.4)
plt.title('CASH_ADVANCE_DEPENDENCY vs PAYMENT_HEALTH_SCORE')
plt.show()
```

```python
# ============================================================
# 8. EDA: 상관관계 히트맵
# ============================================================

corr_cols = [
    'BALANCE',
    'PURCHASES',
    'ONEOFF_PURCHASES',
    'INSTALLMENTS_PURCHASES',
    'CASH_ADVANCE',
    'CREDIT_LIMIT',
    'PAYMENTS',
    'MINIMUM_PAYMENTS',
    'PRC_FULL_PAYMENT',
    'LIMIT_USAGE_RATE',
    'PURCHASE_TO_LIMIT',
    'CASH_ADVANCE_TO_LIMIT',
    'PAYMENT_TO_BALANCE',
    'CASH_ADVANCE_DEPENDENCY',
    'REVOLVING_RISK_SCORE',
    'PAYMENT_HEALTH_SCORE'
]

corr_cols = [col for col in corr_cols if col in df_feature.columns]

plt.figure(figsize=(14, 10))
sns.heatmap(
    df_feature[corr_cols].corr(),
    cmap='coolwarm',
    center=0,
    annot=False
)
plt.title('Correlation Heatmap')
plt.show()
```

```python
# ============================================================
# 9. 모델링용 변수 선택
# ============================================================

base_features = [
    'BALANCE',
    'BALANCE_FREQUENCY',
    'PURCHASES',
    'ONEOFF_PURCHASES',
    'INSTALLMENTS_PURCHASES',
    'CASH_ADVANCE',
    'PURCHASES_FREQUENCY',
    'ONEOFF_PURCHASES_FREQUENCY',
    'PURCHASES_INSTALLMENTS_FREQUENCY',
    'CASH_ADVANCE_FREQUENCY',
    'CASH_ADVANCE_TRX',
    'PURCHASES_TRX',
    'CREDIT_LIMIT',
    'PAYMENTS',
    'MINIMUM_PAYMENTS',
    'PRC_FULL_PAYMENT',
    'TENURE'
]

engineered_features = [
    'LIMIT_USAGE_RATE',
    'PURCHASE_TO_LIMIT',
    'CASH_ADVANCE_TO_LIMIT',
    'PAYMENT_TO_BALANCE',
    'MIN_PAYMENT_PRESSURE',
    'ONEOFF_PURCHASE_RATIO',
    'INSTALLMENT_PURCHASE_RATIO',
    'CASH_ADVANCE_DEPENDENCY',
    'PURCHASE_DEPENDENCY',
    'AVG_PURCHASE_AMOUNT_PER_TRX',
    'AVG_CASH_ADVANCE_AMOUNT_PER_TRX',
    'PURCHASE_ACTIVITY_SCORE',
    'REVOLVING_RISK_SCORE',
    'PAYMENT_HEALTH_SCORE'
]

selected_features = [
    col for col in base_features + engineered_features
    if col in df_feature.columns
]

print('모델링에 사용할 변수 개수:', len(selected_features))
print(selected_features)

# 해석용 데이터: 원본 스케일 유지
df_report = df_feature.copy()

# 모델링용 데이터
model_df = df_feature[selected_features].copy()
model_df = model_df.replace([np.inf, -np.inf], np.nan).fillna(0)

display(model_df.head())
```

```python
# ============================================================
# 10. 이상치 완화, 로그변환, 중복 변수 제거, 표준화
# ============================================================

# 10-1. 이상치 완화: 상하위 0.1% 기준으로 clipping
model_df_clipped = model_df.copy()

for col in model_df_clipped.columns:
    lower = model_df_clipped[col].quantile(0.001)
    upper = model_df_clipped[col].quantile(0.999)
    model_df_clipped[col] = model_df_clipped[col].clip(lower, upper)

# 10-2. 로그변환: 왜도가 큰 양수 변수에만 적용
log_exclude_cols = [
    'BALANCE_FREQUENCY',
    'PURCHASES_FREQUENCY',
    'ONEOFF_PURCHASES_FREQUENCY',
    'PURCHASES_INSTALLMENTS_FREQUENCY',
    'CASH_ADVANCE_FREQUENCY',
    'PRC_FULL_PAYMENT',
    'TENURE',
    'ONEOFF_PURCHASE_RATIO',
    'INSTALLMENT_PURCHASE_RATIO',
    'PURCHASE_ACTIVITY_SCORE'
]

model_df_log = model_df_clipped.copy()
log_cols = []

for col in model_df_log.columns:
    if col not in log_exclude_cols:
        if (model_df_log[col] >= 0).all() and abs(model_df_log[col].skew()) > 1:
            model_df_log[col] = np.log1p(model_df_log[col])
            log_cols.append(col)

print('로그변환 적용 변수:')
print(log_cols)

# 10-3. 상관관계가 지나치게 높은 변수 제거
corr_matrix = model_df_log.corr().abs()
upper_triangle = corr_matrix.where(
    np.triu(np.ones(corr_matrix.shape), k=1).astype(bool)
)

high_corr_drop_cols = [
    col for col in upper_triangle.columns
    if any(upper_triangle[col] > 0.98)
]

model_df_final = model_df_log.drop(columns=high_corr_drop_cols)

print('상관관계가 너무 높아 제거된 변수:')
print(high_corr_drop_cols)
print('최종 모델링 변수 개수:', model_df_final.shape[1])

# 10-4. 표준화
scaler = StandardScaler()
X_scaled = scaler.fit_transform(model_df_final)

X_scaled_df = pd.DataFrame(
    X_scaled,
    columns=model_df_final.columns,
    index=model_df_final.index
)

display(X_scaled_df.head())
```

```python
# ============================================================
# 11. 군집 수 평가: Elbow, Silhouette, DBI, CH Score
# ============================================================

score_list = []
K_RANGE = range(2, 11)

for k in K_RANGE:
    kmeans = KMeans(
        n_clusters=k,
        random_state=RANDOM_STATE,
        n_init=50
    )

    labels = kmeans.fit_predict(X_scaled)

    score_list.append({
        'k': k,
        'inertia': kmeans.inertia_,
        'silhouette': silhouette_score(X_scaled, labels),
        'davies_bouldin': davies_bouldin_score(X_scaled, labels),
        'calinski_harabasz': calinski_harabasz_score(X_scaled, labels)
    })

score_df = pd.DataFrame(score_list)
display(score_df)
```

```python
# ============================================================
# 12. 군집 수 평가 시각화
# ============================================================

plt.figure(figsize=(8, 5))
plt.plot(score_df['k'], score_df['inertia'], marker='o')
plt.title('Elbow Method')
plt.xlabel('Number of Clusters')
plt.ylabel('Inertia')
plt.show()

plt.figure(figsize=(8, 5))
plt.plot(score_df['k'], score_df['silhouette'], marker='o')
plt.title('Silhouette Score')
plt.xlabel('Number of Clusters')
plt.ylabel('Silhouette Score')
plt.show()

plt.figure(figsize=(8, 5))
plt.plot(score_df['k'], score_df['davies_bouldin'], marker='o')
plt.title('Davies-Bouldin Index - lower is better')
plt.xlabel('Number of Clusters')
plt.ylabel('DBI')
plt.show()

plt.figure(figsize=(8, 5))
plt.plot(score_df['k'], score_df['calinski_harabasz'], marker='o')
plt.title('Calinski-Harabasz Score - higher is better')
plt.xlabel('Number of Clusters')
plt.ylabel('CH Score')
plt.show()
```

```python
# ============================================================
# 13. 추천 군집 수 선정
# ============================================================
# 마케팅 전략으로 해석하기 좋은 범위는 보통 3~6개
# Silhouette은 높을수록 좋고, DBI는 낮을수록 좋고, CH Score는 높을수록 좋음

candidate_df = score_df[(score_df['k'] >= 3) & (score_df['k'] <= 6)].copy()

candidate_df['silhouette_rank'] = candidate_df['silhouette'].rank(ascending=False)
candidate_df['dbi_rank'] = candidate_df['davies_bouldin'].rank(ascending=True)
candidate_df['ch_rank'] = candidate_df['calinski_harabasz'].rank(ascending=False)

candidate_df['total_rank'] = (
    candidate_df['silhouette_rank'] +
    candidate_df['dbi_rank'] +
    candidate_df['ch_rank']
)

recommended_k = int(candidate_df.sort_values('total_rank').iloc[0]['k'])

print('지표 기반 추천 군집 수:', recommended_k)
display(candidate_df.sort_values('total_rank'))

# 필요하면 아래 값을 직접 바꿔도 됨
FINAL_K = recommended_k
print('최종 사용할 군집 수:', FINAL_K)
```

```python
# ============================================================
# 14. 최종 K-Means 군집분석 적용
# ============================================================

kmeans_final = KMeans(
    n_clusters=FINAL_K,
    random_state=RANDOM_STATE,
    n_init=50
)

cluster_labels = kmeans_final.fit_predict(X_scaled)

df_report['Cluster'] = cluster_labels

# CUST_ID가 있었다면 다시 붙이기
if id_col is not None and 'CUST_ID' not in df_report.columns:
    df_report.insert(0, 'CUST_ID', id_col.values)

print('최종 군집 수:', FINAL_K)
print('군집별 고객 수')
display(df_report['Cluster'].value_counts().sort_index())
```

```python
# ============================================================
# 15. PCA 2차원 시각화
# ============================================================

pca = PCA(n_components=2, random_state=RANDOM_STATE)
pca_result = pca.fit_transform(X_scaled)

pca_df = pd.DataFrame(pca_result, columns=['PC1', 'PC2'])
pca_df['Cluster'] = cluster_labels

print('PCA 설명 분산 비율:', pca.explained_variance_ratio_)
print('누적 설명 분산:', pca.explained_variance_ratio_.sum())

plt.figure(figsize=(9, 7))
sns.scatterplot(
    data=pca_df,
    x='PC1',
    y='PC2',
    hue='Cluster',
    palette='tab10',
    alpha=0.7,
    s=40
)
plt.title('Customer Segments Visualized by PCA')
plt.show()
```

```python
# ============================================================
# 16. 군집별 고객 수 및 비중
# ============================================================

cluster_size = df_report.groupby('Cluster').size().rename('CUSTOMER_COUNT')
cluster_ratio = (cluster_size / len(df_report) * 100).round(2).rename('CUSTOMER_RATIO_%')

cluster_count_df = pd.concat([cluster_size, cluster_ratio], axis=1)
display(cluster_count_df)

cluster_count_df['CUSTOMER_COUNT'].plot(kind='bar', figsize=(8, 5))
plt.title('Number of Customers by Cluster')
plt.xlabel('Cluster')
plt.ylabel('Customer Count')
plt.show()
```

```python
# ============================================================
# 17. 군집별 프로파일링: 평균/중앙값 테이블
# ============================================================
# 극단값이 많은 금융 데이터는 평균보다 중앙값 해석이 더 안정적임

profile_cols = [
    'BALANCE',
    'PURCHASES',
    'ONEOFF_PURCHASES',
    'INSTALLMENTS_PURCHASES',
    'CASH_ADVANCE',
    'CREDIT_LIMIT',
    'PAYMENTS',
    'MINIMUM_PAYMENTS',
    'PRC_FULL_PAYMENT',
    'TENURE',
    'PURCHASES_TRX',
    'CASH_ADVANCE_TRX',
    'LIMIT_USAGE_RATE',
    'PURCHASE_TO_LIMIT',
    'CASH_ADVANCE_TO_LIMIT',
    'PAYMENT_TO_BALANCE',
    'ONEOFF_PURCHASE_RATIO',
    'INSTALLMENT_PURCHASE_RATIO',
    'CASH_ADVANCE_DEPENDENCY',
    'PURCHASE_DEPENDENCY',
    'REVOLVING_RISK_SCORE',
    'PAYMENT_HEALTH_SCORE'
]

profile_cols = [col for col in profile_cols if col in df_report.columns]

cluster_median_profile = df_report.groupby('Cluster')[profile_cols].median().round(3)
cluster_mean_profile = df_report.groupby('Cluster')[profile_cols].mean().round(3)

cluster_profile = pd.concat([cluster_count_df, cluster_median_profile], axis=1)

print('군집별 중앙값 프로파일')
display(cluster_profile)

print('군집별 평균 프로파일')
display(cluster_mean_profile)
```

```python
# ============================================================
# 18. 군집별 상대적 특징 히트맵
# ============================================================
# 전체 평균 대비 각 군집이 어떤 변수가 높은지/낮은지 확인

key_profile_cols = [
    'BALANCE',
    'PURCHASES',
    'CASH_ADVANCE',
    'CREDIT_LIMIT',
    'PAYMENTS',
    'PRC_FULL_PAYMENT',
    'LIMIT_USAGE_RATE',
    'PURCHASE_TO_LIMIT',
    'CASH_ADVANCE_TO_LIMIT',
    'PAYMENT_TO_BALANCE',
    'CASH_ADVANCE_DEPENDENCY',
    'REVOLVING_RISK_SCORE'
]

key_profile_cols = [col for col in key_profile_cols if col in df_report.columns]

cluster_mean_key = df_report.groupby('Cluster')[key_profile_cols].mean()

z_profile = (cluster_mean_key - df_report[key_profile_cols].mean()) / df_report[key_profile_cols].std()

plt.figure(figsize=(14, 6))
sns.heatmap(
    z_profile,
    cmap='coolwarm',
    center=0,
    annot=True,
    fmt='.1f'
)
plt.title('Cluster Profile Heatmap - Relative to Overall Average')
plt.show()
```

```python
# ============================================================
# 19. 군집별 주요 변수 막대그래프
# ============================================================

plot_cols = [
    'PURCHASES',
    'CASH_ADVANCE',
    'BALANCE',
    'CREDIT_LIMIT',
    'PAYMENTS',
    'LIMIT_USAGE_RATE',
    'PURCHASE_TO_LIMIT',
    'CASH_ADVANCE_DEPENDENCY',
    'PRC_FULL_PAYMENT',
    'REVOLVING_RISK_SCORE'
]

plot_cols = [col for col in plot_cols if col in df_report.columns]

for col in plot_cols:
    cluster_median = df_report.groupby('Cluster')[col].median()

    plt.figure(figsize=(8, 5))
    cluster_median.plot(kind='bar')
    plt.title(f'Median {col} by Cluster')
    plt.xlabel('Cluster')
    plt.ylabel(col)
    plt.show()
```

```python
# ============================================================
# 20. 군집명 자동 부여
# ============================================================
# 자동 분류는 초안용이며, 최종 보고서에서는 프로파일 테이블을 보고 군집명을 수정해도 됨

q30 = df_report[profile_cols].quantile(0.30)
q60 = df_report[profile_cols].quantile(0.60)
q70 = df_report[profile_cols].quantile(0.70)

cluster_median_for_label = df_report.groupby('Cluster')[profile_cols].median()

def assign_segment_name(row):
    # 현금서비스 의존 고객
    if (
        row.get('CASH_ADVANCE', 0) >= q70.get('CASH_ADVANCE', 0) and
        row.get('CASH_ADVANCE_DEPENDENCY', 0) >= q70.get('CASH_ADVANCE_DEPENDENCY', 0)
    ):
        return '현금서비스 의존 고객'

    # 고구매 우량 고객
    elif (
        row.get('PURCHASES', 0) >= q70.get('PURCHASES', 0) and
        row.get('PAYMENTS', 0) >= q60.get('PAYMENTS', 0) and
        row.get('PRC_FULL_PAYMENT', 0) >= q60.get('PRC_FULL_PAYMENT', 0)
    ):
        return '고구매 우량 고객'

    # 잔액 부담 고객
    elif (
        row.get('BALANCE', 0) >= q70.get('BALANCE', 0) and
        row.get('LIMIT_USAGE_RATE', 0) >= q70.get('LIMIT_USAGE_RATE', 0) and
        row.get('PRC_FULL_PAYMENT', 0) <= q30.get('PRC_FULL_PAYMENT', 0)
    ):
        return '잔액 부담 고객'

    # 할부 중심 고객
    elif (
        row.get('INSTALLMENT_PURCHASE_RATIO', 0) >= q70.get('INSTALLMENT_PURCHASE_RATIO', 0) and
        row.get('INSTALLMENTS_PURCHASES', 0) >= q60.get('INSTALLMENTS_PURCHASES', 0)
    ):
        return '할부 중심 고객'

    # 저활성 고객
    elif (
        row.get('PURCHASES', 0) <= q30.get('PURCHASES', 0) and
        row.get('PURCHASES_TRX', 0) <= q30.get('PURCHASES_TRX', 0) and
        row.get('CASH_ADVANCE', 0) <= q30.get('CASH_ADVANCE', 0)
    ):
        return '저활성 고객'

    else:
        return '일반 혼합 고객'

segment_name_map = {}

for cluster_id, row in cluster_median_for_label.iterrows():
    segment_name_map[cluster_id] = assign_segment_name(row)

df_report['Segment_Name'] = df_report['Cluster'].map(segment_name_map)

segment_summary = (
    df_report
    .groupby(['Cluster', 'Segment_Name'])
    .size()
    .reset_index(name='CUSTOMER_COUNT')
)

segment_summary['CUSTOMER_RATIO_%'] = (
    segment_summary['CUSTOMER_COUNT'] / len(df_report) * 100
).round(2)

display(segment_summary)
```

```python
# ============================================================
# 21. 세그먼트별 마케팅 전략 정리
# ============================================================

strategy_map = {
    '고구매 우량 고객': [
        '프리미엄 카드 업그레이드 제안',
        '포인트 적립률 강화',
        '항공, 호텔, 쇼핑 제휴 혜택 제공',
        '신용한도 상향 제안 가능'
    ],

    '할부 중심 고객': [
        '무이자 할부 프로모션 제공',
        '가전, 여행, 교육 업종 제휴 혜택 제공',
        '장기 할부 수수료 할인',
        '고가 상품 구매 캠페인 타겟팅'
    ],

    '현금서비스 의존 고객': [
        '현금서비스 수수료 부담 안내',
        '저금리 대체 금융상품 제안',
        '상환 계획 알림 서비스 제공',
        '리스크 관리 대상 고객으로 모니터링'
    ],

    '잔액 부담 고객': [
        '분할납부 전환 안내',
        '연체 방지 알림 제공',
        '최소결제 반복 고객 관리',
        '무리한 한도 증액보다 상환 유도 중심 전략'
    ],

    '저활성 고객': [
        '첫 결제 캐시백 제공',
        '편의점, 카페, 배달앱 등 생활업종 할인',
        '앱 푸시 쿠폰 발송',
        '휴면 방지 캠페인 진행'
    ],

    '일반 혼합 고객': [
        '사용 카테고리 기반 맞춤 혜택 제공',
        '월별 소비 리포트 제공',
        '카드 사용 빈도 증가 캠페인',
        '포인트 적립 이벤트 제공'
    ]
}

strategy_rows = []

for cluster_id in sorted(df_report['Cluster'].unique()):
    segment_name = segment_name_map[cluster_id]
    customer_count = (df_report['Cluster'] == cluster_id).sum()
    customer_ratio = customer_count / len(df_report) * 100

    strategy_rows.append({
        'Cluster': cluster_id,
        'Segment_Name': segment_name,
        'Customer_Count': customer_count,
        'Customer_Ratio_%': round(customer_ratio, 2),
        'Marketing_Strategy': ' / '.join(strategy_map[segment_name])
    })

strategy_df = pd.DataFrame(strategy_rows)
display(strategy_df)
```

```python
# ============================================================
# 22. 보고서 작성용 문장 자동 출력
# ============================================================

for _, row in strategy_df.iterrows():
    print('=' * 80)
    print(f"Cluster {row['Cluster']} - {row['Segment_Name']}")
    print(f"고객 수: {row['Customer_Count']}명")
    print(f"고객 비중: {row['Customer_Ratio_%']}%")
    print()

    temp_profile = cluster_median_profile.loc[row['Cluster']]

    print('[주요 특징]')
    for col in [
        'BALANCE',
        'PURCHASES',
        'CASH_ADVANCE',
        'CREDIT_LIMIT',
        'PAYMENTS',
        'PRC_FULL_PAYMENT',
        'LIMIT_USAGE_RATE',
        'PURCHASE_TO_LIMIT',
        'CASH_ADVANCE_DEPENDENCY',
        'REVOLVING_RISK_SCORE'
    ]:
        if col in temp_profile.index:
            print(f'- {col}: {temp_profile[col]}')

    print()
    print('[마케팅 전략]')
    for strategy in strategy_map[row['Segment_Name']]:
        print(f'- {strategy}')
    print()
```
